If the Bill amount is 1000, not equal to 1000, less than 1000, then they are offering 10%.
If the bill amount is 1000 or more but less than 5000, then they are offering 15% discount on it and
from 5000 to 10,000 they are offering 20% discount.
And beyond 10,000 they are offering 25% discount.
Input is Bill Amount.

# All

```python
bill = float(input("Enter the Billing Amount"))
if bill <1000:
    print(" Congrats you got 10% discount of", bill*.1,' Amount to pay', bill-bill*.1)
elif bill >=1000 and bill<=5000:
    print(" Congrats you got 15% discount of", bill*.15,' Amount to pay', bill-bill*.15)
elif bill >=5000 and bill <=10000:
    print(" Congrats you got 20% discount of", bill*.2,' Amount to pay', bill-bill*.2)
else:
     print(" Congrats you got 25% discount of", bill*.25,' Amount to pay', bill-bill*.25)
    
```

# Count the digits of Number

```python
n = int(input("Enter the number"))
count =0
while(n>0):
    count=count+1;
    n=n//10

print('Count is', count)
```
