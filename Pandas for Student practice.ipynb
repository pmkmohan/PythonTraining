# Pandas Brackets: When to Use Which One?
    Pandas uses normal Python syntax. The main symbols you will use are:
    Symbol	Common purpose
    ()	     Call a function or method; pass arguments
    []	     Select data, index data, or slice data
    :	     Specify a range/slice inside []
    {}	     Create dictionaries or sets
    .	     Access a method or attribute
