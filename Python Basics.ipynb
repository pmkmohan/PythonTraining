#functions

```python
def is_even(n):
    return n%2 ==0
nums = [3,2,6,8,4,6,2,9]
evens = list(filter(is_even,nums))
```

```python
nums
```

```python
evens
```

```python
evens1= list(filter(lambda n:n%2==0, nums))
print(evens1)
```

```python
odds= list(filter(lambda n:n%2!=0, nums))
print(odds)
```

```python
def update(n):
    return n+2
nums = [3,2,6,8,4,6,2,9]
evens = list(filter(is_even,nums))
double = list(map(update,evens))
```

```python
nums,evens,double
```

```python
doubles = list(map(lambda n:n+2,nums))
print(doubles)
```

```python
from functools import reduce
def add_all(a,b):
    return a+b
nums = [3,2,6,8,4,6,2,9]
evens = list(filter(is_even,nums))
double = list(map(update,evens))
redu= reduce(add_all,evens)
redu1 = reduce(add_all,double)
```

```python
print(nums)
print(evens)
print(redu)
print(redu1)
```

```python
sum= (reduce(lambda a,b:a+b,double))
sum
```
