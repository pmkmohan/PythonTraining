# Sets

    1) Unordered & Unindexed collection of items. 
    2) Set elements are unique. Duplicate elements are not allowed. 
    3) Set elements are immutable (cannot be changed). 
    4) Set itself is mutable. We can add or remove items from it.
    5) set is given {}

 # Dictionary 
    Dictionary is a mutable data type in Python. 
    A python dictionary is a collection of key and value pairs separated by a colon (:) & enclosed in curly braces {}. 
    Keys must be unique in a dictionary, duplicate values are allowed. 

```python
mydict = dict() # empty dictionary 
```

```python
mydict
```

```python

mydict = {1:'one' , 2:'two' , 3:'three'} # dictionary with integer keys 
mydict
```

```python
mydict[1]
```

```python
mydict1 = dict({1:'one' , 2:'two' , 3:'three'}) # Create dictionary using dict() 
mydict1
```

```python
mydict2 = {'A':'one' , 'B':'two' , 'C':'three'} # dictionary with character keys 
mydict2
```

```python
mydict = {1:'one' , 'A':'two' , 3:'three'} # dictionary with mixed keys 
mydict 
```

```python
mydict.keys() # Return Dictionary Keys using keys() method 
```

```python
mydict2 =mydict.copy()
mydict2
```

```python
mydict2.values()  # Return Dictionary Values using values() method
```

```python
mydict2.items()  # Access each key-value pair within a dictionary
```
